```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}は大人です")
    else:
      print(f"{self.name}は大人ではありません")

human1 = Human("テセウス", 26)
human2 = Human("ニュート", 19)
human3 = Human("ティナ", 21)
human4 = Human("クイニー", 18)
human5 = Human("アルバス", 99)

human_list = [human1, human2, human3, human4, human5]

for human in human_list:
  human.check_adult()
```
